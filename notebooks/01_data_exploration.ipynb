````python
# En Pandas, el equivalente más directo a un UNION ALL de SQL es pd.concat()
# Pandas concat = pegar estructuras: axis=0 hacia abajo; axis=1 hacia el costado.
# ignore_index=True hace que Pandas genere nuevamente el índice de filas 0, 1, 2, 3... para el DataFrame unificado

df_all = pd.concat(
    [df_2009_2010, df_2010_2011],
    ignore_index=True
)
````

````python
df_all[(df_all['Description'].notna()) & (df_all['Customer ID'].isna())]
````

````python
# Guardo el filtro en un DataFrame para no repetirlo constantemente

df_sin_customer = df_all[
    (df_all['Customer ID'].isna()) &
    (df_all['Description'].notna())
]
````

````python
df_all.duplicated().sum()

# Me está diciendo:

# Encontré 34.335 filas que son exactamente iguales a una fila anterior considerando todas las columnas
````

````python
# vemos un par de casos duplicados para ver si son realmente iguales o si hay alguna diferencia en alguna columna
df_all[df_all.duplicated()].head(10)
````

````python
# Marcame todas las filas que tengan otra idéntica, incluida la primera.
df_duplicados = df_all[
    df_all.duplicated(keep=False)
]

# Podemos ordenar por Invoice y StockCode
df_duplicados = df_duplicados.sort_values(
    by=['Invoice', 'StockCode']
)

# vemos el output de los 20 primeros registros duplicados
df_duplicados.head(20)
````

Se detectaron **34.335 filas duplicadas**, pero no se eliminan porque no existe un identificador de línea (`Line Item`) que permita saber si son duplicados reales o varias unidades del mismo producto registradas en líneas separadas. Eliminarlas podría provocar una pérdida de ventas válidas y subestimar cantidades e importes.
